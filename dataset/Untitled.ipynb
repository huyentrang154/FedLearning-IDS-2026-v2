```python
import numpy as np
import pandas as pd
# dir_path = "nslkdd/"
# train_data = np.genfromtxt(f'{dir_path}/rawdata/NSLKDD_Train.csv', delimiter=",")
# test_data = np.genfromtxt(f'{dir_path}/rawdata/NSLKDD_Test.csv', delimiter=",")
# data = np.concatenate((train_data, test_data), axis=0)
# dir_path = "ctu13_13/"
# data = np.genfromtxt(f'{dir_path}/rawdata/CTU13_13.csv', delimiter=",")
dir_path = "cic_ids/"
train_data = pd.read_csv(f'{dir_path}/rawdata/CIC-IDS-2017-Train-10percent.csv', sep=",", header=None)
test_data = pd.read_csv(f'{dir_path}/rawdata/CIC-IDS-2017-Test-10percent.csv', sep=',', header=None)

# drop nan value
train_data.replace([np.inf, -np.inf], np.nan, inplace=True)
train_data.dropna(inplace=True)
test_data.replace([np.inf, -np.inf], np.nan, inplace=True)
test_data.dropna(inplace=True)

# concatenate
data = pd.concat([train_data, test_data])
data.reset_index(drop=True, inplace=True)
```

```python
df = pd.DataFrame(data)
df.head()
df.shape
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Tính số lượng giá trị cho cột 'Category'
value_counts = data[85].value_counts()

# Vẽ biểu đồ cột
plt.figure(figsize=(8, 6))
value_counts.plot(kind='barh')

# Thêm tiêu đề và nhãn cho các trục
plt.title('Thống kê số lượng bản ghi của từng lớp dữ liệu của tập CIC-IDS2017')
plt.xlabel('Loại nhãn')
plt.ylabel('Số lượng')
plt.xticks(rotation=360)
# Hiển thị đồ thị
plt.savefig('NSL-KDD.png', bbox_inches='tight')
```
